# Lab 2: Probability - Simple Prediction

This notebook has two simple ways to predict something. `most_common_label` makes a lazy guess from what happened most often. `linear_regression_fit` and `predict` fit a straight line `y = slope * x + intercept` to some points and use it to predict a number. `mean_squared_error` tells you how wrong the predictions are, so you can check whether the line really beats the lazy guess. The reading panel explains the math step by step.

Spreadsheets use this same method for a trendline. There is no training loop, just a few sums. It is the simplest example of "fit a model to data". Labs 4 and 5 show models that learn step by step instead.


**You get:** two lists of numbers with the same length, like `x = [0, 1, 2]` and `y = [1, 3, 5]`. Position `i` in both lists is one point. (`zip(a, b)` walks two lists side by side: `(a[0], b[0])`, then `(a[1], b[1])`, and so on.)
**You give back:** a pair `(slope, intercept)` of floats: `(2.0, 1.0)`.
*Linear regression* finds the straight line `y = slope * x + intercept` that sits closest to the points. The **slope** is how much `y` goes up when `x` goes up by 1. The **intercept** is the value of `y` where `x` is 0. "Closest" means that the squared gaps between the points and the line, added up, are as small as possible. This is called *least squares*. The formulas below give that line directly.
**Idea, in words:**
- `x_mean` and `y_mean` are the averages of the two lists. The *mean* (average) is `sum(values) / len(values)`.
- `slope` = (sum of `(x_i - x_mean) * (y_i - y_mean)`) divided by (sum of `(x_i - x_mean) ** 2`).
- `intercept` = `y_mean - slope * x_mean`.

**Check (by hand):** `x = [0, 1, 2]`, `y = [1, 3, 5]`: `x_mean = 1`, `y_mean = 3`. Top: `(-1)(-2) + (0)(0) + (1)(2) = 4`. Bottom: `1 + 0 + 1 = 2`. Slope = `4 / 2 = 2.0`, intercept = `3 - 2.0 * 1 = 1.0`. So the answer is `(2.0, 1.0)`. Another: `x = [1, 2, 3]`, `y = [1, 3, 2]` gives `(0.5, 1.0)`.

## Solution

````python
def most_common_label(labels: list[int]) -> int:
    counts: dict[int, int] = {}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    return max(counts, key=lambda label: counts[label])


def linear_regression_fit(x: list[float], y: list[float]) -> tuple[float, float]:
    n = len(x)
    x_mean = sum(x) / n
    y_mean = sum(y) / n
    numerator = sum((xi - x_mean) * (yi - y_mean) for xi, yi in zip(x, y))
    denominator = sum((xi - x_mean) ** 2 for xi in x)
    slope = numerator / denominator
    intercept = y_mean - slope * x_mean
    return slope, intercept


def predict(x: float, slope: float, intercept: float) -> float:
    return slope * x + intercept


def mean_squared_error(y_true: list[float], y_pred: list[float]) -> float:
    n = len(y_true)
    return sum((yt - yp) ** 2 for yt, yp in zip(y_true, y_pred)) / n


print(linear_regression_fit(x_toy, y_toy))
````
