# Lab 2: Probability - Simple Prediction

This notebook implements two "simple prediction" strategies: `most_common_label` (a frequency baseline) and `linear_regression_fit` / `predict` (fitting a straight line `y = slope * x + intercept` with the least-squares closed form). `mean_squared_error` is how we'll judge whether the regression actually beats the baseline. The full walkthrough of the regression math and the baseline concept is in the reading panel.

This is the same closed-form fit spreadsheets use for a trendline - no training loop, just algebra - and it's the simplest example of "fit a model to data" before labs 4-5 introduce models that need iterative training instead.

## Exercise

### Goal
Build two simple ways to predict something. First a **frequency guess** ("just say whatever happened most often"). Then a **straight-line model** that predicts a number `y` from a number `x`, plus a way to score how wrong the predictions are. All four functions are used again in the project.

### Words you will see
- **Label**: the right answer for an example. Here 0 or 1 (for instance negative / positive review).
- **Baseline**: a very simple guess that more clever models are compared against.
- **Linear regression**: finding the straight line `y = slope * x + intercept` that sits closest to a set of `(x, y)` points. The **slope** is how much `y` goes up when `x` goes up by 1. The **intercept** is the value of `y` where `x` is 0.
- **Least squares**: the rule for "closest": make the sum of squared vertical gaps between the points and the line as small as possible. The formulas below give that line directly.
- **Mean**: the average, `sum(values) / len(values)`.
- **Prediction**: the `y` the model says goes with a given `x`.
- **Mean squared error (MSE)**: the average of `(true - predicted) ** 2`. Squaring removes the sign and punishes big misses more. 0 is perfect, lower is better.
- **`zip(a, b)`**: walks two lists side by side, giving `(a[0], b[0])`, then `(a[1], b[1])`, and so on.
- **`max(d, key=...)`**: the item that scores highest under `key`. `max(d, key=lambda k: d[k])` gives the key of `d` with the biggest value.

### 1. `most_common_label(labels)`
**Receives:** a `list[int]` of 0/1 labels, e.g. `[1, 0, 1, 1, 0]`.
**Returns:** one `int`, the label that appears most often: `1`. On a tie the label that appeared first in the list wins (`[1, 0]` gives `1`, `[0, 1]` gives `0`).
**Idea:** count each label, then pick the label with the biggest count.
**Check:** `[0, 0, 1]` has two 0s and one 1, so the answer is `0`.

### 2. `linear_regression_fit(x, y)`
**Receives:** two lists of numbers of the same length, e.g. `x = [0, 1, 2]`, `y = [1, 3, 5]`. Position `i` in both lists is one point.
**Returns:** a tuple `(slope, intercept)` of floats: `(2.0, 1.0)`.
**Idea, in words:**
- `x_mean` and `y_mean` are the averages of the two lists.
- `slope` = (sum of `(x_i - x_mean) * (y_i - y_mean)`) divided by (sum of `(x_i - x_mean) ** 2`).
- `intercept` = `y_mean - slope * x_mean`.

**Check (by hand):** `x = [0, 1, 2]`, `y = [1, 3, 5]`: `x_mean = 1`, `y_mean = 3`. Top: `(-1)(-2) + (0)(0) + (1)(2) = 4`. Bottom: `1 + 0 + 1 = 2`. Slope = `4 / 2 = 2.0`, intercept = `3 - 2.0 * 1 = 1.0`. So the answer is `(2.0, 1.0)`. Another: `x = [1, 2, 3]`, `y = [1, 3, 2]` gives `(0.5, 1.0)`.

### 3. `predict(x, slope, intercept)`
**Receives:** a **single** number `x` and the line's `slope` and `intercept`.
**Returns:** one float, `slope * x + intercept`.
**Idea:** put `x` into the line equation.
**Check:** `predict(6, 2.0, 0.0)` is `12.0`, and `predict(10, 0.5, 1.0)` is `6.0`.

### 4. `mean_squared_error(y_true, y_pred)`
**Receives:** two lists of numbers of the same length: the real values and the predictions, e.g. `[2, 4]` and `[3, 3]`.
**Returns:** one float, the average of the squared differences: `1.0`.
**Idea:** for each pair take `true - predicted`, square it, add up all the squares, divide by how many there are.
**Check:** `[2, 4]` vs `[3, 3]`: differences `-1` and `1`, squares `1` and `1`, average `1.0`. Also `[1, 2, 3]` vs `[2, 2, 5]`: squares `1, 0, 4`, sum 5, average `5/3`, about `1.667`.

## Solution

```python
def most_common_label(labels: list[int]) -> int:
    counts: dict[int, int] = {}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    return max(counts, key=lambda label: counts[label])


def linear_regression_fit(x: list[float], y: list[float]) -> tuple[float, float]:
    n = len(x)
    x_mean = sum(x) / n
    y_mean = sum(y) / n
    numerator = sum((xi - x_mean) * (yi - y_mean) for xi, yi in zip(x, y))
    denominator = sum((xi - x_mean) ** 2 for xi in x)
    slope = numerator / denominator
    intercept = y_mean - slope * x_mean
    return slope, intercept


def predict(x: float, slope: float, intercept: float) -> float:
    return slope * x + intercept


def mean_squared_error(y_true: list[float], y_pred: list[float]) -> float:
    n = len(y_true)
    return sum((yt - yp) ** 2 for yt, yp in zip(y_true, y_pred)) / n


print(linear_regression_fit(x_toy, y_toy))
```
