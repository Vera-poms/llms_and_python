### Goal
A perceptron is the smallest "decision maker" in machine learning. It takes a few numbers, mixes them into one score, and answers with a plain yes or no (`+1` or `-1`). The class below already stores the settings (`weights`, `bias`) and already computes the score (`net_input`). Your job is to finish the last two steps: turn the score into `+1` / `-1`, and chain the steps together.

### Words you will see
- **Perceptron**: a tiny classifier. It multiplies each input by a weight, adds them up, adds a bias, and picks a side.
- **Feature vector `x`**: the list of numbers describing one example, stored as a numpy array. `np.array([1.0, 0.0])` is an example with 2 features.
- **Weights**: one number per feature saying how much that feature matters (a negative weight pushes the answer toward `-1`). Stored in `self.weights`.
- **Bias**: one extra number added to the score no matter what the inputs are. Stored in `self.bias`.
- **Net input `z`**: the score. `z = (x[0]*w[0] + x[1]*w[1] + ...) + bias`. The method `net_input(x)` already returns it as a float.
- **Activation**: the rule that turns the score `z` into a final class label.
- **Class label**: the answer. Here it is the integer `+1` or `-1` (not `0`, not `True`/`False`).
- **Method / `self`**: a function that lives inside a class. `self` is the perceptron object itself, so `self.net_input(x)` calls another method on the same object.
- **Assert**: a line that stops the program with an error message if the condition is false.

### `activation(self, z)`
**Receives:** `z`, a single number (a `float`) - the score from `net_input`.

**Must return:** the integer `1` if the score is zero or above, otherwise the integer `-1`. Nothing else may come out of it.

**Idea in plain words:** positive side or zero means yes (`+1`), negative side means no (`-1`). Be careful with exactly `0`: the rule says "`z >= 0`", so `0` counts as `+1`.

**Check:**
- `activation(0.3)` is `1`
- `activation(-0.3)` is `-1`
- `activation(0.0)` is `1` (the edge case: zero goes to the `+1` side)

### `predict(self, x)`
**Receives:** `x`, a numpy array holding one example's features, e.g. `np.array([1.0, 0.0])`.

**Must return:** `1` or `-1` (an `int`) - the perceptron's answer for that example.

**Idea in plain words:** two steps in a row. First compute the score from `x` with the method that already exists. Then feed that score into your `activation`. Use `self.` to call both; do not rewrite the dot product.

**Check:** with `weights = [1.0, -1.0]` and `bias = 0.5`:
- `x = [1.0, 0.0]`: score = `1*1 + (-1)*0 + 0.5 = 1.5`, which is `>= 0`, so the answer is `1`.
- `x = [0.0, 2.0]`: score = `1*0 + (-1)*2 + 0.5 = -1.5`, which is below 0, so the answer is `-1`.

A brand-new `Perceptron(2)` has all weights 0 and bias 0, so every score is `0.0` and it answers `1` for every input.

## Solution

```python
class Perceptron:
    def __init__(self, n_features: int):
        self.weights = np.zeros(n_features)
        self.bias = 0.0

    def net_input(self, x: np.ndarray) -> float:
        return float(np.dot(x, self.weights) + self.bias)

    def activation(self, z: float) -> int:
        return 1 if z >= 0 else -1

    def predict(self, x: np.ndarray) -> int:
        return self.activation(self.net_input(x))


p = Perceptron(n_features=2)
p.weights = np.array([1.0, -1.0])
p.bias = 0.5
print(p.predict(np.array([1.0, 0.0])))
```

```python
MIN_WORD_COUNT = 3
MIN_POLARITY_RATIO = 0.7

y = np.array([1 if label == 1 else -1 for label in labels])


def tok(sentence: str) -> list[str]:
    return re.findall(r"[a-z]+", sentence.lower())


def build_vocab(sentences: list[str], y: np.ndarray) -> list[str]:
    """Pick words whose presence strongly leans toward one label."""
    pos_doc_counts: dict[str, int] = {}
    neg_doc_counts: dict[str, int] = {}
    for sentence, label in zip(sentences, y):
        bucket = pos_doc_counts if label == 1 else neg_doc_counts
        for word in set(tok(sentence)):
            bucket[word] = bucket.get(word, 0) + 1

    vocab = []
    for word in set(pos_doc_counts) | set(neg_doc_counts):
        pos, neg = pos_doc_counts.get(word, 0), neg_doc_counts.get(word, 0)
        total = pos + neg
        if total < MIN_WORD_COUNT:
            continue
        if abs(pos - neg) / total < MIN_POLARITY_RATIO:
            continue
        vocab.append((word, total))

    vocab.sort(key=lambda pair: pair[1], reverse=True)
    return [word for word, _ in vocab]


def vectorize(sentences: list[str], vocab: list[str]) -> np.ndarray:
    """Binary bag-of-words: 1 if the vocab word appears in the sentence, else 0."""
    index = {word: i for i, word in enumerate(vocab)}
    X = np.zeros((len(sentences), len(vocab)))
    for row, s in enumerate(sentences):
        for w in set(tok(s)):
            if w in index:
                X[row, index[w]] = 1.0
    return X


def pca_2d(X: np.ndarray) -> np.ndarray:
    centered = X - X.mean(axis=0)
    cov = np.cov(centered, rowvar=False)
    eigenvalues, eigenvectors = np.linalg.eigh(cov)
    top2 = eigenvectors[:, np.argsort(eigenvalues)[::-1][:2]]
    return centered @ top2


def plot_boundary(X2d: np.ndarray, y: np.ndarray, model: Perceptron, title: str) -> None:
    plt.figure(figsize=(7, 6))
    for label, color, name in ((1, "seagreen", "positive"), (-1, "crimson", "negative")):
        mask = y == label
        plt.scatter(X2d[mask, 0], X2d[mask, 1], color=color, alpha=0.7, label=name)

    x_min, x_max = X2d[:, 0].min() - 1, X2d[:, 0].max() + 1
    if abs(model.weights[1]) > 1e-9:
        xs = np.linspace(x_min, x_max, 100)
        ys = -(model.weights[0] * xs + model.bias) / model.weights[1]
        plt.plot(xs, ys, color="black", linestyle="--", label="decision boundary")

    plt.xlabel("PCA component 1")
    plt.ylabel("PCA component 2")
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.show()


vocab = build_vocab(sentences, y)
X = vectorize(sentences, vocab)
X2d = pca_2d(X)

model = Perceptron(n_features=2)
predictions = np.array([model.predict(x) for x in X2d])
accuracy = (predictions == y).mean()

print(f"Vocabulary size: {len(vocab)}")
print(f"Untrained perceptron accuracy on real sentiment data: {accuracy:.2%}")
print("(Weights are all zero, so it predicts the same class for everything - see lab 5.)")

plot_boundary(X2d, y, model, "Untrained perceptron on real review sentiment")
```
